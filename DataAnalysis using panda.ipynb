## TASK 4

### LAMDA FUNCTIONS WITH MAP , FILTER ,REDUCE:

#### 01.MAP 
* LAMBDA - CREATE FUNCTION, DOES NOT PROCESS DATA
* MAP -    USE FUNCTION, PROCESS THE DATA


```python
import pandas as pd
map_file = pd.read_csv("insurance.csv")

female_smoker_list = (map_file["sex"].map(lambda x: x == "female") & map_file["smoker"].map(lambda x: x == "yes"))

print("Female smokers:", female_smoker_list)
```

```python
map_file.shape
```

#### 02.FILTER
* FILTER - ONLY GIVE ANSWER THE CONDITION MATCHES
* ZIP - USED FOR COMBINE BOTH TABLES

```python
import pandas as pd
filter_FILE = pd.read_csv("insurance.csv")

female_smokers = list(filter( lambda row: row[0] == "female" and row[1] == "yes",zip(filter_FILE["sex"], filter_FILE["smoker"])))

print("Female smokers:",female_smokers)
```

#### 03.REDUCE

* REDUCE - ITS ONLY GIVE TOTAL 

```python
import pandas as pd
from functools import reduce

insurance_file = pd.read_csv("insurance.csv")
max_bmi = reduce(lambda x, y: x if x > y else y, insurance_file["bmi"])

print("Maximum BMI:", max_bmi)
```
